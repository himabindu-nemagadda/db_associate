## Creating Delta Lake Tables

**Note:** If your workspace does not support the `hive_metastore` catalog, switch to the **unity-catalog** branch in this Git Folder.

```sql
USE CATALOG hive_metastore
```

```sql
CREATE TABLE employees
  (id INT, name STRING, salary DOUBLE);
```


## Catalog Explorer

Check the created **employees** table in the **Catalog** explorer.

## Inserting Data

```sql
INSERT INTO employees
VALUES 
  (1, "Adam", 3500.0),
  (2, "Sarah", 4020.5);

INSERT INTO employees
VALUES
  (3, "John", 2999.3),
  (4, "Thomas", 4000.3);

INSERT INTO employees
VALUES
  (5, "Anna", 2500.0);

INSERT INTO employees
VALUES
  (6, "Kim", 6200.3)
```

```sql
SELECT * FROM employees
```

## Updating Table

```sql
UPDATE employees 
SET salary = salary + 100
WHERE name LIKE "A%"
```
